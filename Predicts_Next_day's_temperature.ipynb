```python
import numpy as np
from sklearn.linear_model import LinearRegression

data = [20, 22, 21, 24, 26, 25, 28, 30, 29, 32]

# Number of previous values used for prediction
window = 3

X = []
y = []

for i in range(len(data) - window):
    X.append(data[i:i + window])
    y.append(data[i + window])

X = np.array(X)
y = np.array(y)

# Train model
model = LinearRegression()
model.fit(X, y)

# Use last 3 values to predict next value
last_values = np.array(data[-window:]).reshape(1, -1)

prediction = model.predict(last_values)

print("Last 3 values:", data[-window:])
print("Predicted next value:", round(prediction[0], 2))
```
